```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(-np.pi, np.pi, 35)
```

```python
def f(x):
  return np.sin(x)
```

```python
y = f(x)
```

```python
plt.plot(x,y)
```

```python
def DerivativeR(f, x, h=1e-2):
  return (f(x+h) - f(x))/h
```

```python
def DerivativeL(f, x, h=1e-2):
  return (f(x) - f(x-h))/h
```

```python
def DerivativeC(f, x, h=1e-2):
  return (f(x+h) - f(x-h))/(2*h)
```

```python
DerivativeR(f,0.)
```

```python
DerivativeR(f,90./180*np.pi)
```

```python
# Queremos pintar
DR = DerivativeR(f,x)
DL = DerivativeR(f,x)
DC = DerivativeC(f,x)
```

```python
plt.plot(x,DR,label='Derecha')
plt.plot(x,DL,label='Izquierda')
plt.plot(x,DC,label='Izquierda')

plt.plot(x,np.cos(x),label='Real')
plt.legend()
```

```python
plt.plot(x,np.abs(DR-np.cos(x)),label='Real 1')
plt.plot(x,np.abs(DL-np.cos(x)),label='Real 2')
plt.plot(x,np.abs(DC-np.cos(x)),label='Real Central')
plt.legend()
```

```python
x= np.linspace(0,10,50)
```

```python
def funcion(x):
  return 3*x+1
```

```python
def pit(x,y):
  return np.sqrt((x-4)**2 + (y-5)**2)
```

```python
rta=1000
for i in range(len(x)):
  if pit(x[i],y[i])<rta:
    rta=pit(x[i],y[i])
rta
```
